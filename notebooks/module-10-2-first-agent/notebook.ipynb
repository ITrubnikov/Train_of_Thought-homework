## 1. Свой `@tool` — контракт: типы + docstring с `Args`

smolagents строит описание инструмента для модели из аннотаций типов и docstring.
Без секции `Args` модель не поймёт параметры.

```python
@tool
def get_current_time_in_timezone(timezone: str) -> str:
    """Возвращает текущее локальное время в часовом поясе.

    Args:
        timezone: IANA-имя пояса, например 'Europe/London'.
    """
    import datetime, zoneinfo
    return datetime.datetime.now(zoneinfo.ZoneInfo(timezone)).strftime("%H:%M")
```

## 2. Собираем `CodeAgent` и запускаем

Одна строка модели, одна — агента. Петлю think→act→observe крутит `run()`.

```python
model = InferenceClientModel(model_id="Qwen/Qwen2.5-Coder-32B-Instruct")
agent = CodeAgent(tools=[get_current_time_in_timezone], model=model)
print(agent.run("Который сейчас час в Лондоне?"))
```
